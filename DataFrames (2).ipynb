```python
data = {
    'Student': ['Aman', 'Riya', 'Rahul', 'Priya', 'Karan'],
    'Course': ['Python', 'Java', 'Python', 'SQL', 'Java'],
    'Score': [85, 90, 78, 88, 92]
}

course_df = pd.DataFrame(data)

# Display column names
print(course_df.columns)

# Retrieve Student
print(course_df['Student'])

# Create DataFrame with Student and Score
new_df = course_df[['Student', 'Score']]

print(new_df)

# Add Batch column
course_df['Batch'] = '2026'

print(course_df)
```

```python
data = {
    'Name': ['Aman', 'Riya', 'Rahul', 'Priya', 'Karan'],
    'Place': ['Delhi', 'Mumbai', 'Jaipur', 'Agra', 'Pune']
}

travel_df = pd.DataFrame(data)

# Add Code using np.arange(5)
travel_df['Code'] = np.arange(5)

print(travel_df)

# Create Series
codes = pd.Series(
    [100, 400],
    index=[1, 4]
)

# Replace Code column
travel_df['Code'] = codes

print(travel_df)
```

```python
data = {
    'Student': ['Aman', 'Riya', 'Rahul', 'Priya'],
    'Class': [2, 2, 2, 2],
    'Section': ['A', 'A', 'B', 'B']
}

school_df = pd.DataFrame(data)

new_df = school_df.reindex(
    columns=['Student', 'Class', 'Section', 'House']
)

print(new_df)

# Assign common value to House
new_df['House'] = 'Blue'

print(new_df)
```

```python
data = {
    'Product': ['Laptop', 'Phone', 'Tablet', 'Watch', 'Camera'],
    'Price': [50000, 20000, 30000, 5000, 40000]
}

inventory_df = pd.DataFrame(data)

# Partially indexed Series
stock = pd.Series(
    [10, 20, 15],
    index=[0, 2, 4]
)

# Add column
inventory_df['Stock'] = stock

print(inventory_df)

# Delete Stock column
del inventory_df['Stock']

print(inventory_df)
```
