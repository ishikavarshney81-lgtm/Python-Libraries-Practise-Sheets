```python
#Q35
import numpy as np

np.random.seed(42)

marks = np.random.randint(20, 101, size=(12, 4))

print("Marks:\n", marks)

student_avg = np.mean(marks, axis=1)
test_mean = np.mean(marks, axis=0)
test_std = np.std(marks, axis=0)

print("Student Averages:", student_avg)
print("Test Means:", test_mean)
print("Test Standard Deviations:", test_std)

passed_all = np.all(marks >= 40, axis=1)
print("Students passing all tests:", np.sum(passed_all))

print("Students with average >= 75:",
      np.where(student_avg >= 75)[0] + 1)

print("Test with greatest variation:",
      np.argmax(test_std) + 1)
```

```python
#Q36
import numpy as np

a = np.arange(21, 46).reshape(5, 5)

print("Original Matrix:\n", a)

central = a[1:4, 1:4]
print("Central 3x3:\n", central)

transposed = central.T
print("Transpose:\n", transposed)

product = np.matmul(central, transposed)
print("Product:\n", product)

row_sums = np.sum(product, axis=1)
print("Row-wise Sums:", row_sums)

maximum = np.max(product)
positions = np.argwhere(product == maximum)

print("Maximum:", maximum)
print("Positions (zero-based):\n", positions)
```

```python
#Q37
import numpy as np

quantity = np.array([
    [10, 20, 15, 8],
    [12, 18, 20, 10],
    [15, 25, 18, 12],
    [20, 22, 25, 15],
    [18, 20, 22, 18],
    [25, 28, 30, 20]
])

original_prices = np.array([150, 240, 90, 320])

daily_revenue = np.sum(quantity * original_prices, axis=1)
product_quantity = np.sum(quantity, axis=0)
product_revenue = product_quantity * original_prices

print("Daily Revenue:", daily_revenue)
print("Total Revenue per Product:", product_revenue)
print("Highest Revenue Day:", np.argmax(daily_revenue) + 1)
print("Product Quantities:", product_quantity)
print("Best-selling Product:", np.argmax(product_quantity) + 1)

new_prices = np.where(product_quantity > 100,
                     original_prices * 0.90,
                     original_prices)

new_total_revenue = np.sum(quantity * new_prices)

print("Original Prices:", original_prices)
print("Reduced Prices:", new_prices)
print("New Total Revenue:", new_total_revenue)
```

```python
#Q38
import numpy as np

np.random.seed(42)

data = np.random.normal(50, 12, 100)

outside = (data < 30) | (data > 70)
print("Values outside range:", np.sum(outside))

cleaned = np.where(data < 30, 30,
                   np.where(data > 70, 70, data))

print("\nOriginal Statistics:")
print("Mean:", np.mean(data))
print("Variance:", np.var(data))
print("Standard Deviation:", np.std(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))

print("\nCleaned Statistics:")
print("Mean:", np.mean(cleaned))
print("Variance:", np.var(cleaned))
print("Standard Deviation:", np.std(cleaned))
print("Minimum:", np.min(cleaned))
print("Maximum:", np.max(cleaned))
```

```python
#Q39
import pandas as pd

branch_a = pd.Series({'Q1': 120, 'Q2': 145, 'Q4': 180})
branch_b = pd.Series({'Q2': 110, 'Q3': 135, 'Q4': 155})
branch_c = pd.Series({'Q1': 95, 'Q3': 125, 'Q4': 140})

ordinary = branch_a + branch_b + branch_c
print("Ordinary Addition:")
print(ordinary)

common = branch_a.index.intersection(branch_b.index)
common = common.intersection(branch_c.index)

print("Common Quarters:", common.tolist())
print("Combined Earnings in Common Quarters:",
      ordinary.loc[common])

zero_result = (
    branch_a.add(branch_b, fill_value=0)
            .add(branch_c, fill_value=0)
)

print("Missing Treated as Zero:")
print(zero_result)
```

```python
#Q40
import pandas as pd

scores = {
    'R101': 78,
    'R102': 65,
    'R104': 92,
    'R105': 48
}

required = ['R105', 'R103', 'R101', 'R106', 'R104', 'R102']

s = pd.Series(scores, index=required)

print("Original Series:")
print(s)

print("\nMissing Student Records:")
print(s[s.isnull()])

valid = s[s.notnull() & (s >= 70)]

print("\nValid Scores >= 70:")
print(valid)

d = valid.to_dict()
print("\nDictionary:", d)

reconstructed = pd.Series(d)
reconstructed = reconstructed.reindex(valid.index)

print("\nReconstructed Series:")
print(reconstructed)

print("\nValues Match:", valid.equals(reconstructed))
```
