```python
import pandas as pd
from pandas import Series

sales = Series(
    [250, 750, 400, 900],
    index=['Pen', 'Bag', 'Book', 'Laptop']
)

print("Values:")
print(sales.values)

print("\nIndex:")
print(sales.index)

print("\nValue of Bag:")
print(sales['Bag'])

print("\nSales above 500:")
print(sales[sales > 500])

print("\nIs Book present?")
print('Book' in sales.index)

print("\nDictionary:")
print(sales.to_dict())
```
